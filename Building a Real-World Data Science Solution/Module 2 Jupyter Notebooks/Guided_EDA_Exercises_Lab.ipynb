```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def analyze_ticket_patterns(df):
    """
    Perform exploratory data analysis on TicketWise ticket routing patterns
    
    Parameters:
        df (pd.DataFrame): TicketWise dataset
        
    Returns:
        dict: Dictionary containing analysis results and statistical summaries
    """
    # -------------------------------
    # 🕒 Convert submission_time to datetime
    # -------------------------------
    df['submission_time'] = pd.to_datetime(df['submission_time'], errors='coerce')
    
    # Ensure chronological sorting
    df = df.sort_values('submission_time')
    
    # -------------------------------
    # 📈 1. Daily ticket submissions over time
    # -------------------------------
    daily_counts = df.groupby(df['submission_time'].dt.date).size()
    
    plt.figure(figsize=(12, 5))
    plt.plot(daily_counts.index, daily_counts.values, color='royalblue', linewidth=2)
    plt.title('Daily Ticket Submissions Over Time', fontsize=14)
    plt.xlabel('Date')
    plt.ylabel('Number of Tickets')
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.tight_layout()
    plt.show()
    
    # 💬 Insight:
    # - This line chart reveals workload trends over time.
    # - Look for spikes that might correspond to product releases, outages, or Monday morning surges.

    # -------------------------------
    # 📊 2. Boxplot of resolution times by subscription tier
    # -------------------------------
    plt.figure(figsize=(8, 5))
    sns.boxplot(data=df, x='subscription_tier', y='resolution_time', palette='coolwarm')
    plt.title('Resolution Time Distribution by Subscription Tier', fontsize=14)
    plt.xlabel('Subscription Tier')
    plt.ylabel('Resolution Time (hours)')
    plt.tight_layout()
    plt.show()
    
    # 💬 Insight:
    # - Enterprise tiers often show faster resolutions due to SLA commitments.
    # - Wider spread in "Basic" may suggest inconsistent service handling.

    # -------------------------------
    # 📋 3. Basic statistics of key columns
    # -------------------------------
    key_cols = [
        'attachments', 'previous_interactions', 'contract_value',
        'account_age_months', 'open_tickets', 'linked_ticket_count',
        'avg_resolution_customer', 'avg_resolution_tier', 'resolution_time'
    ]
    
    summary_stats = df[key_cols].describe().T  # transpose for readability
    
    # 💬 Insight:
    # - High variance in `contract_value` may indicate diverse customer base sizes.
    # - Correlate `previous_interactions` with `is_urgent` later to study frustration buildup.

    # -------------------------------
    # 📈 4. Optional correlation heatmap (bonus visualization)
    # -------------------------------
    plt.figure(figsize=(10, 6))
    sns.heatmap(df[key_cols].corr(), cmap='coolwarm', annot=True, fmt=".2f")
    plt.title('Correlation Heatmap of Numeric Features', fontsize=14)
    plt.tight_layout()
    plt.show()

    # 💬 Insight:
    # - Strong correlation between avg_resolution_customer and avg_resolution_tier implies 
    #   that customer history aligns closely with tier expectations.
    # - Negative correlation between contract_value and resolution_time suggests
    #   high-value clients receive faster support.

    # -------------------------------
    # 📦 Return structured analysis output
    # -------------------------------
    results = {
        "daily_ticket_counts": daily_counts,
        "summary_statistics": summary_stats,
        "notable_insights": [
            "Ticket submissions may spike around key business days or product events.",
            "Enterprise customers tend to have lower resolution times due to priority handling.",
            "High variance in resolution times among Basic users suggests potential for process optimization."
        ]
    }
    
    return results
```

```python
import pandas as pd
import numpy as np
import boto3
from io import StringIO
from sklearn.preprocessing import OneHotEncoder

def preprocess_ticket_data(df, output_bucket):
    """
    Clean and preprocess TicketWise ticket data
    
    Parameters:
        df (pd.DataFrame): Raw TicketWise dataset
        output_bucket (str): Target S3 bucket name (e.g., 'ticketwise-processed-data')
        
    Returns:
        (pd.DataFrame, bool): Tuple containing:
            - Processed DataFrame
            - Boolean upload status (True if S3 upload succeeded)
    """
    # -------------------------------
    # 🧹 1. Handle Missing Values
    # -------------------------------
    # Replace missing categorical fields with 'Unknown'
    cat_cols = ['channel', 'subscription_tier', 'issue_type']
    df[cat_cols] = df[cat_cols].fillna('Unknown')
    
    # Fill missing numeric fields with median to avoid skew
    num_cols = df.select_dtypes(include=['int64', 'float64']).columns
    for col in num_cols:
        df[col] = df[col].fillna(df[col].median())
    
    # -------------------------------
    # 🕒 2. Extract Day of Week from submission_time
    # -------------------------------
    df['submission_time'] = pd.to_datetime(df['submission_time'], errors='coerce')
    df['day_of_week'] = df['submission_time'].dt.dayofweek  # Monday=0, Sunday=6
    
    # Replace any NaT-derived NaNs with -1 for consistency
    df['day_of_week'] = df['day_of_week'].fillna(-1).astype(int)

    # 💬 Insight:
    # Day-of-week patterns can highlight workload clustering (e.g., Monday peaks, weekend lulls).

    # -------------------------------
    # 🔢 3. Standardize / Encode Categorical Fields
    # -------------------------------
    encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
    
    encoded = encoder.fit_transform(df[cat_cols])
    encoded_cols = encoder.get_feature_names_out(cat_cols)
    encoded_df = pd.DataFrame(encoded, columns=encoded_cols, index=df.index)
    
    # Drop original categorical columns and merge encoded ones
    df_processed = pd.concat([df.drop(columns=cat_cols), encoded_df], axis=1)

    # 💬 Insight:
    # One-hot encoding ensures models can interpret non-numeric category data effectively.

    # -------------------------------
    # 🧾 4. Optional Feature Normalization
    # -------------------------------
    # You can apply scaling later in the ML pipeline;
    # for now, numeric data remains in raw scale to keep interpretability.

    # -------------------------------
    # ☁️ 5. Upload to S3
    # -------------------------------
    success = False
    try:
        # Convert DataFrame to CSV in-memory
        csv_buffer = StringIO()
        df_processed.to_csv(csv_buffer, index=False)
        
        # Initialize S3 client
        s3 = boto3.client('s3')
        filename = 'processed_ticketwise_data.csv'
        
        s3.put_object(
            Bucket=output_bucket,
            Key=filename,
            Body=csv_buffer.getvalue()
        )
        success = True
        print(f"✅ Successfully uploaded processed data to s3://{output_bucket}/{filename}")
    except Exception as e:
        print(f"⚠️ S3 upload failed: {e}")
    
    # -------------------------------
    # 📦 6. Return processed DataFrame and upload status
    # -------------------------------
    return df_processed, success
```
