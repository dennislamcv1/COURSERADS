## Task 1: Initial Setup and Data Loading Solution Code

<b>Steps:</b> 

```python
# Import required libraries
import pandas as pd
import numpy as np
from datetime import datetime
import boto3

# Read the uploaded csv file (make sure it is saved in the same folder as the notebook)
df = pd.read_csv('ticketwise_dataset_uncleaned.csv')

# Display initial data quality issues
print("Missing values:")
print(df.isnull().sum())
print("\nSample of inconsistent categories:")
print(df['channel'].value_counts())
```

## Task 2: Deriving day_of_week Solution Code


<b>Steps:</b> 

```python
# Convert submission_time to datetime
df['submission_time'] = pd.to_datetime(df['submission_time'])

# 1. Extract the weekday name (like Monday, Sunday) from 'submission_time' using .dt.day_name()
df['day_of_week'] = df['submission_time'].dt.day_name()

# 2. Print some values to check the weekday names are correct
print(df['day_of_week'].head())

# 3. Show how many tickets were submitted each weekday by counting their occurrences
print(df['day_of_week'].value_counts().sort_index())
```

## Task 3: Standardizing Categorical Fields Solution Code


<b>Steps:</b> 

```python
# 1. Clean and standardize channel
df['channel'] = df['channel'].str.lower().str.strip()
df['channel'] = df['channel'].replace('web_form', 'web form')

# 2. Clean subscription_tier
df['subscription_tier'] = df['subscription_tier'].str.title().str.strip()
df['subscription_tier'] = df['subscription_tier'].fillna('Unknown')

# 3. Verify the changes
print("Cleaned categories:")
print(df['channel'].value_counts())
print(df['subscription_tier'].value_counts())
```

## Task 4: Handling Missing Values and Text Cleaning Solution Code 


<b>Steps:</b> 

```python
# 1. Clean issue_type
df['issue_type'] = df['issue_type'].str.title().str.strip()
df['issue_type'] = df['issue_type'].fillna('Other')

# 2. Clean product_area
df['product_area'] = df['product_area'].str.title().str.strip()
df['product_area'] = df['product_area'].fillna('General')

# 3. Display cleaning results
print("Cleaned text fields:")
print(df['issue_type'].value_counts())
print(df['product_area'].value_counts())
```

## Task 5: Final Validation and Storage Solution Code


<b>Steps:</b> 

```python
# 1. Perform final data quality checks
print("Checking for missing values:")
print(df.isnull().sum())

# 2. Upload the cleaned DataFrame directly to S3 as a CSV string
import io
import boto3

csv_buffer = io.StringIO()
df.to_csv(csv_buffer, index=False)

s3 = boto3.client("s3")
bucket_name = "your-bucket-name"  # change this to your bucket name
s3_key = "cleaned_data/ticket_data_cleaned.csv"

# 3. Print messages to confirm if the upload was successful or if there were any errors
try:
    s3.put_object(Bucket=bucket_name, Key=s3_key, Body=csv_buffer.getvalue())
    print(f"File uploaded to s3://{bucket_name}/{s3_key}")
except Exception as e:
    print("Error uploading file:", e)
```
