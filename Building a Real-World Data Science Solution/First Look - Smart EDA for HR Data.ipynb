## Initial Data Exploration

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('employee_insights.csv') 
```

```python
print("First few rows:")
df.head()
```

```python
print("\nDataset information:")
df.info()
```

## Descriptive Statistics

```python
df['salary'] = df['salary'].str.replace("$","").astype(float)   
df['salary'] = df['salary'].fillna(df['salary'].mean())
print(df['salary'].head())
```

```python
print("Summary statistics for numerical columns:")
numerical_stats = df[['salary', 'hours_worked_weekly', 'satisfaction_score']].describe() 
print(numerical_stats)
```

```python
print("\nDepartment distribution:")
print(df['department'].value_counts())
```

```python
df['department'] = df['department'].str.capitalize()
df['department'].value_counts() 
```

## Pattern Identification

```python
# Group by analysis
dept_stats = df.groupby('department').agg({'satisfaction_score': ['mean', 'count'],
                                           'salary': 'mean'}).round(2) 

print("Department Analysis:")
print(dept_stats)
```

```python
df['work_mode'] = df['work_mode'].str.capitalize()  
df['work_mode'] = df['work_mode'].str.replace("Remote work", "Remote")

work_mode_stats = df.groupby('work_mode')['satisfaction_score'].describe()

print("\nSatisfaction by Work Mode:")
print(work_mode_stats)
```

## Outlier Detection

```python
def identify_outliers(df, column):  

    Q1 = df[column].quantile(0.25)  
    Q3 = df[column].quantile(0.75)  
    IQR = Q3 - Q1                   
    lower_bound = Q1 - 1.5 * IQR    
    upper_bound = Q3 + 1.5 * IQR    
    
    outliers = df[(df[column] < lower_bound) | 
                  (df[column] > upper_bound)]
    
    return outliers
```

```python
salary_outliers = identify_outliers(df, 'salary')  
print("Salary outliers found:", len(salary_outliers))  
print("\nOutlier Details:")
print(salary_outliers[['department', 'salary']].head())
```
