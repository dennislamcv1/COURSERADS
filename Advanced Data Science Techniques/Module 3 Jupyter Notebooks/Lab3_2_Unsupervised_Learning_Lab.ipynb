## Task 1: Data Preparation Solution Code


<b>Step 1:</b> Load and Examine Data

```python
# Import required libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('insightlysoft_dataset.csv')

# Display first few rows and basic information
print(df.head())

print("\nDataset Info:")
print(df.info())
```

<b>Step 2:</b>  Select and Scale Features

```python
# Select relevant numeric features for clustering
features_for_clustering = [
    'monthly_login_freq',
    'num_support_tickets',
    'avg_product_usage_hours',
    'num_users',
    'nps_score',
    'feature_adoption_score',
    'customer_satisfaction_score'
]

# Create feature matrix with the selected features
X = df[features_for_clustering]

# Scale the features using StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## Task 2: Implementing K-means Clustering Solution Code

<b>Step 1:</b> Determine Optimal Number of Clusters

```python
# Prepare to store inertia values for different numbers of clusters
inertia_values = []

# Define the range of k values to test (from 1 to 10 clusters)
k_values = range(1, 11)

# Loop through each k value and fit KMeans
for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertia_values.append(kmeans.inertia_)

# Plot elbow curve
plt.figure(figsize=(10, 6))
plt.plot(k_values, inertia_values, 'bx-')
plt.xlabel('k')
plt.ylabel('Inertia')
plt.title('Elbow Method For Optimal k')
plt.show()
```

<b>Step 2:</b> Apply K-means Clustering

```python
# Apply k-means with optimal k
optimal_k = 5  # Adjust based on elbow curve

# Create a KMeans model using the chosen number of clusters
kmeans = KMeans(n_clusters=optimal_k, random_state=42)

# Fit the model to the scaled data and get cluster labels for each point
cluster_labels = kmeans.fit_predict(X_scaled)

# Add cluster labels to original dataframe
df['Cluster'] = cluster_labels
```

## Task 3: Dimensionality Reduction with PCA Solution Code

<b>Step 1:</b> Apply PCA

```python
# Apply PCA to reduce dimensions to 3 components
pca = PCA(n_components=3)
X_pca = pca.fit_transform(X_scaled)

# Calculate explained variance ratio
explained_variance = pca.explained_variance_ratio_
print("\nExplained variance ratio:", explained_variance)
```

<b>Step 2:</b> Visualize Clusters in PCA Space

```python
# Create 3D scatter plot of clusters
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
scatter = ax.scatter(X_pca[:, 0], X_pca[:, 1], X_pca[:, 2],
                     c=cluster_labels,
                     cmap='viridis',
                     s=50, alpha=0.8)
ax.set_title('Customer Segments in 3D PCA Space')
ax.set_xlabel('PC 1')
ax.set_ylabel('PC 2')
ax.set_zlabel('PC 3')

# Add color bar for clusters
fig.colorbar(scatter, ax=ax, label='Cluster')
plt.show()
```

## Task 4: Analyzing Cluster Characteristics Solution Code

<b>Step 1:</b> Calculate Cluster Profiles

```python
# Calculate mean values for each cluster
cluster_profiles = df.groupby('Cluster')[features_for_clustering].mean()
print("\nCluster Profiles:")
print(cluster_profiles)

# Calculate cluster sizes
cluster_sizes = df['Cluster'].value_counts()
print("\nCluster Sizes:")
print(cluster_sizes)
```

<b>Step 2:</b> Visualize Key Metrics by Cluster

```python
# Create boxplots for key metrics
key_metrics = ['monthly_login_freq', 'avg_product_usage_hours', 'customer_satisfaction_score']

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for i, metric in enumerate(key_metrics):
    df.boxplot(column=metric, by='Cluster', ax=axes[i])
    axes[i].set_title(f'{metric} by Cluster')
plt.tight_layout()
plt.show()
```
