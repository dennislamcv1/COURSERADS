## Task 2: Time Series Visualization Solution Code

<b>Step 1:</b> Basic Time Series Analysis

```python
# Create visualizations to understand business performance
def plot_business_metrics(df):
    """
    Create subplot showing revenue and user engagement trends
    Parameters:
        df: DataFrame with date index
    """
    # Set figure size and layout
    fig, axs = plt.subplots(2, 1, figsize=(12, 8), sharex=True)

    # Plot revenue and costs
    axs[0].plot(df.index, df['revenue'], label='Revenue', color='green')
    axs[0].plot(df.index, df['costs'], label='Costs', color='red')
    axs[0].set_title('Revenue vs Costs Over Time')
    axs[0].set_ylabel('Amount ($)')
    axs[0].legend()
    axs[0].grid(True)

    # Plot user engagement metrics
    axs[1].plot(df.index, df['unique_users'], label='Unique Users', color='blue')
    axs[1].plot(df.index, df['leads'], label='Leads', color='orange')
    axs[1].set_title('User Engagement Over Time')
    axs[1].set_ylabel('Count')
    axs[1].set_xlabel('Date')
    axs[1].legend()
    axs[1].grid(True)

    # Show the plots
    plt.tight_layout()
    plt.show()

plot_business_metrics(df)
```

<b> Step 2:</b> Moving Average Implementation

```python
# Calculate moving averages for key metrics
def calculate_moving_averages(df, window=7):
    """
    Calculate moving averages for revenue and users
    Parameters:
        df: DataFrame with metrics
        window: Rolling window size (default 7 days)
    """
    df = df.copy()  # Avoid modifying the original DataFrame

    # Calculate moving averages
    df['revenue_ma'] = df['revenue'].rolling(window=window).mean()
    df['unique_users_ma'] = df['unique_users'].rolling(window=window).mean()

    return df

df_with_ma = calculate_moving_averages(df, window=7)
print(df_with_ma[['revenue', 'revenue_ma', 'unique_users', 'unique_users_ma']].head(10))
```

## Task 4: Forecasting and Model Evaluation Solution Code
Build and evaluate a forecasting model using prophet  for InsightlySoft's revenue data using industry-standard metrics.

<b>Step 1:</b>  Prepare Data

```python
def prepare_forecast_data(df):
    """
    Split time series data into training and test sets for propeht model
    Returns: X_train, X_test, y_train, y_test
    """
    # Ensure index is datetime
    df.index = pd.to_datetime(df.index)

    # Reset index and rename columns for Prophet
    prophet_df = df.reset_index()[['date', 'revenue']].rename(columns={'date': 'ds', 'revenue': 'y'})

    # Check columns
    print("Columns in prophet_df:", prophet_df.columns)
    print(prophet_df.head())

    # Split 80% train, 20% test
    split_index = int(len(prophet_df) * 0.8)
    train_df = prophet_df.iloc[:split_index]
    test_df = prophet_df.iloc[split_index:]

    # Select features and target separately
    X_train = train_df[['ds']]
    y_train = train_df['y']
    X_test = test_df[['ds']]
    y_test = test_df['y']
    return X_train, X_test, y_train, y_test

X_train, X_test, y_train, y_test = prepare_forecast_data(df)
```

<b>Step 2:</b> Build Model

```python
def build_forecast_model(X_train, y_train):
    """
    Create and train prophet forecasting model
    Returns: Trained model and predictions
    """
    # Combine X and y into a single DataFrame as Prophet expects
    train_df = pd.concat([X_train, y_train], axis=1)

    # Initialize and fit the model
    model = Prophet()
    model.fit(train_df)

    # Create a future DataFrame with same dates as training data
    future = model.make_future_dataframe(periods=0)  # No future steps yet, just fitting
    forecast = model.predict(future)

    return model, forecast

model, forecast_train = build_forecast_model(X_train, y_train)
# To forecast future values:
future = model.make_future_dataframe(periods=len(X_test))  # For test period
forecast_test = model.predict(future)
```

<b>Step 3:</b> Calculate Metrics

```python
def calculate_metrics(y_true, y_pred):
    """
    Calculate MAE, RMSE, and MAPE
    Returns: Dictionary of metrics
    """
    metrics = {
        'mae': np.mean(np.abs(y_true - y_pred)),
        'rmse': np.sqrt(np.mean((y_true - y_pred)**2)),
        'mape': np.mean(np.abs((y_true - y_pred) / y_true)) * 100
    }
    return metrics
```

<b> Step 4: </b> Visualize Results

```python
def plot_forecast_results(y_true, y_pred, metrics):
    """
    Plot actual vs predicted values with metrics
    """
    plt.figure(figsize=(12, 6))
    # Plot actual vs predicted
    plt.plot(y_true.index, y_true.values, label='Actual', marker='o')
    plt.plot(y_true.index, y_pred.values, label='Predicted', marker='x')

    # Add title and labels
    plt.title('Forecast: Actual vs Predicted')
    plt.xlabel('Date')
    plt.ylabel('Value')
    plt.legend()
    plt.grid(True)

    # Display metrics on the plot
    metrics_text = f"MAE: {metrics['mae']}\nRMSE: {metrics['rmse']}\nMAPE: {metrics['mape']}%"
    plt.gcf().text(0.75, 0.7, metrics_text, fontsize=12, bbox=dict(facecolor='white', alpha=0.6))

    plt.tight_layout()
    plt.show()

# Align y_test index to the datetime from X_test (Prophet 'ds' column)
y_test.index = X_test['ds'].values

# Ensure forecast_test has 'ds' as index for alignment
forecast_test = forecast_test.set_index('ds')

# Select predicted values corresponding to y_test dates
y_pred = forecast_test.loc[y_test.index, 'yhat']

# Calculate metrics (assuming calculate_metrics is defined)
metrics = calculate_metrics(y_test, y_pred)

# Plot results
plot_forecast_results(y_test, y_pred, metrics)
```
