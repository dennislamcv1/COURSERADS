## Task 1: Feature Engineering Solution Code
Create meaningful features that capture patterns in urban incidents.

<b>Step 1:</b> Create Time-Based Features

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv('smart_city_incidents_cleaned.csv')
df['timestamp'] = pd.to_datetime(df['timestamp'])

# Create time-based features that might affect incident patterns
def create_temporal_features(df):
    """
    Create features based on timestamp information
    Returns DataFrame with new features
    """
    # 1. Extract basic time components
    df['day_of_week'] = df['timestamp'].dt.dayofweek  # Monday=0, Sunday=6
    df['month'] = df['timestamp'].dt.month

    # 2. Flag rush hour (assume 7–9am and 4–6pm as rush hours)
    df['hour'] = df['timestamp'].dt.hour
    df['is_rush_hour'] = df['hour'].isin([7, 8, 9, 16, 17, 18]).astype(int)

    # 3. Time since last similar incident (group by incident type if available)
    if 'incident_type' in df.columns:
        df = df.sort_values(['incident_type', 'timestamp'])
        df['time_since_last'] = df.groupby('incident_type')['timestamp'].diff().dt.total_seconds()
    else:
        df = df.sort_values('timestamp')
        df['time_since_last'] = df['timestamp'].diff().dt.total_seconds()

    # Fill missing values caused by diff
    df['time_since_last'] = df['time_since_last'].fillna(0)

    return df

df = create_temporal_features(df)
```

<b>Step 2:</b> Create Location-Based Features

```python
# Create features that capture neighborhood patterns
def create_spatial_features(df):
    """
    Create features based on location information
    Returns DataFrame with new features
    """
    # 1. Neighborhood incident rate
    incident_counts = df['neighborhood'].value_counts(normalize=True)
    df['incident_rate_neighborhood'] = df['neighborhood'].map(incident_counts)

    # 2. Average resolution time by neighborhood
    if 'resolution_time' in df.columns:
        avg_resolution = df.groupby('neighborhood')['resolution_time'].mean()
        df['avg_resolution_neighborhood'] = df['neighborhood'].map(avg_resolution)

    # 3. Escalation rate by neighborhood
    if 'escalated' in df.columns:
        escalation_rate = df.groupby('neighborhood')['escalated'].mean()
        df['escalation_rate_neighborhood'] = df['neighborhood'].map(escalation_rate)

    return df

df = create_spatial_features(df)
```

## Task 2: Model Development Solution Code
Build models to predict both escalation needs and resolution times.

<b>Step 1:</b> Prepare Features for Modeling

```python
# Create final feature set for both classification (required_escalation)
# and regression (time_to_resolution) tasks
def prepare_feature_sets(df):
    """
    Prepare separate feature sets for each prediction task.

    Returns:
        X_class: Features for classification
        X_reg: Features for regression
        y_class: Target for classification
        y_reg: Target for regression
    """
    # 1. Define target variables
    y_class = df['required_escalation']  # Binary target for escalation
    y_reg = df['time_to_resolution']     # Continuous target for resolution time

    # 2. Select feature columns (exclude targets and identifiers)
    exclude_cols = ['report_id', 'required_escalation', 'time_to_resolution', 'timestamp']
    features = df.drop(columns=exclude_cols, errors='ignore')

    # 3. Handle categorical variables with one-hot encoding
    categorical_cols = features.select_dtypes(include=['object', 'category']).columns
    features_encoded = pd.get_dummies(features, columns=categorical_cols, drop_first=True)

    # 4. Return features and targets (no scaling needed for Random Forest)
    X_class = features_encoded.copy()
    X_reg = features_encoded.copy()

    return X_class, X_reg, y_class, y_reg

# Prepare features and targets
X_class, X_reg, y_class, y_reg = prepare_feature_sets(df)

# Check shapes to confirm
print("Classification features shape:", X_class.shape)
print("Regression features shape:", X_reg.shape)
print("Classification target shape:", y_class.shape)
print("Regression target shape:", y_reg.shape)
```

<b>Step 2:</b> Build and Evaluate Models

```python
# Implement both classification and regression models (RandomForestClassifier, RandomForestRegressor)
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             mean_squared_error, r2_score, f1_score, roc_auc_score, roc_curve)
import matplotlib.pyplot as plt

def build_and_evaluate_models(X_class, X_reg, y_class, y_reg):
    """
    Build and evaluate both models.

    Returns:
        models: dict with trained 'classifier' and 'regressor'
        metrics: dict with performance metrics for both tasks
    """
    models = {}
    metrics = {}

    # 1. Split data into train/test sets (70% train, 30% test)
    Xc_train, Xc_test, yc_train, yc_test = train_test_split(
        X_class, y_class, test_size=0.3, random_state=42)
    Xr_train, Xr_test, yr_train, yr_test = train_test_split(
        X_reg, y_reg, test_size=0.3, random_state=42)

    # 2. Train RandomForestClassifier
    clf = RandomForestClassifier(random_state=42)
    clf.fit(Xc_train, yc_train)
    models['classifier'] = clf

    # Train RandomForestRegressor
    reg = RandomForestRegressor(random_state=42)
    reg.fit(Xr_train, yr_train)
    models['regressor'] = reg

    # 3. Calculate classification metrics
    y_pred_class = clf.predict(Xc_test)
    y_proba_class = clf.predict_proba(Xc_test)[:, 1]

    metrics['classification'] = {
        'accuracy': accuracy_score(yc_test, y_pred_class),
        'precision': precision_score(yc_test, y_pred_class),
        'recall': recall_score(yc_test, y_pred_class),
        'f1_score': f1_score(yc_test, y_pred_class),
        'roc_auc': roc_auc_score(yc_test, y_proba_class)
    }

    # Calculate regression metrics
    y_pred_reg = reg.predict(Xr_test)
    mse = mean_squared_error(yr_test, y_pred_reg)
    rmse = mse ** 0.5
    r2 = r2_score(yr_test, y_pred_reg)

    metrics['regression'] = {
        'rmse': rmse,
        'r2_score': r2
    }

    # 4. Visualizations

    # Classification ROC Curve
    fpr, tpr, _ = roc_curve(yc_test, y_proba_class)
    plt.figure(figsize=(6, 5))
    plt.plot(fpr, tpr, color='darkorange', lw=2,
             label=f'ROC curve (AUC = {metrics["classification"]["roc_auc"]:.2f})')
    plt.plot([0, 1], [0, 1], color='navy', lw=1, linestyle='--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('Classification ROC Curve')
    plt.legend(loc='lower right')
    plt.grid(True)
    plt.show()

    # Regression Actual vs Predicted
    plt.figure(figsize=(6, 5))
    plt.scatter(yr_test, y_pred_reg, alpha=0.5)
    plt.plot([yr_test.min(), yr_test.max()], [yr_test.min(), yr_test.max()], 'r--')
    plt.xlabel('Actual Resolution Time')
    plt.ylabel('Predicted Resolution Time')
    plt.title('Regression: Actual vs Predicted')
    plt.grid(True)
    plt.show()

    return models, metrics

# Run the function with prepared data
models, metrics = build_and_evaluate_models(X_class, X_reg, y_class, y_reg)

# Print classification metrics
print("Classification metrics:")
for metric, value in metrics['classification'].items():
    print(f"{metric}: {value:.4f}")

# Print regression metrics
print("\nRegression metrics:")
for metric, value in metrics['regression'].items():
    print(f"{metric}: {value:.4f}")
```

## Task 3: Model Optimization Solution Code
Optimize your models while staying within computational constraints.

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import roc_auc_score, mean_squared_error

def optimize_models(models, X_class, X_reg, y_class, y_reg, n_iter=20, cv=3, random_state=42):
    """
    Tune hyperparameters for both models
    Returns optimized models and improvement metrics
    """
    optimized_models = {}
    improvement_metrics = {}

    # 1. Define parameter grids

    param_dist_clf = {
        'n_estimators': [50, 100, 200],
        'max_depth': [None, 10, 20],
        'min_samples_split': [2, 5, 10],
    }

    param_dist_reg = {
        'n_estimators': [50, 100, 200],
        'max_depth': [None, 10, 20],
        'min_samples_split': [2, 5, 10],
    }

    # 2. RandomizedSearchCV for classifier
    clf_random = RandomizedSearchCV(
        estimator=models['classifier'],
        param_distributions=param_dist_clf,
        n_iter=n_iter,
        cv=cv,
        scoring='roc_auc',
        random_state=random_state,
        n_jobs=-1
    )
    clf_random.fit(X_class, y_class)
    optimized_models['classifier'] = clf_random.best_estimator_

    # 3. RandomizedSearchCV for regressor
    reg_random = RandomizedSearchCV(
        estimator=models['regressor'],
        param_distributions=param_dist_reg,
        n_iter=n_iter,
        cv=cv,
        scoring='neg_root_mean_squared_error',  # still correct scoring here
        random_state=random_state,
        n_jobs=-1
    )
    reg_random.fit(X_reg, y_reg)
    optimized_models['regressor'] = reg_random.best_estimator_

    # 4. Evaluate improvements

    # Baseline predictions
    y_proba_baseline = models['classifier'].predict_proba(X_class)[:, 1]
    y_pred_baseline_reg = models['regressor'].predict(X_reg)

    # Tuned predictions
    y_proba_tuned = optimized_models['classifier'].predict_proba(X_class)[:, 1]
    y_pred_tuned_reg = optimized_models['regressor'].predict(X_reg)

    improvement_metrics['classification'] = {
        'baseline_roc_auc': roc_auc_score(y_class, y_proba_baseline),
        'tuned_roc_auc': roc_auc_score(y_class, y_proba_tuned)
    }

    improvement_metrics['regression'] = {
        'baseline_rmse': mean_squared_error(y_reg, y_pred_baseline_reg) ** 0.5,
        'tuned_rmse': mean_squared_error(y_reg, y_pred_tuned_reg) ** 0.5
    }

    return optimized_models, improvement_metrics

optimized_models, improvement_metrics = optimize_models(models, X_class, X_reg, y_class, y_reg)
```

## Task 4: Results Analysis and Recommendations Solution Code

Analyze your results and prepare recommendations for SmartCity Solutions.

```python
from sklearn.inspection import permutation_importance
def create_analysis_report(models, metrics, features, X_class, y_class, X_reg, y_reg):
    """
    Generate report with insights and recommendations.
    """
    report = {}

    # 1. Summarize model performance
    report["performance_summary"] = {
        "Classification ROC AUC":
            f"{metrics['classification']['baseline_roc_auc']:.3f} → {metrics['classification']['tuned_roc_auc']:.3f}",
        "Regression RMSE":
            f"{metrics['regression']['baseline_rmse']:.2f} → {metrics['regression']['tuned_rmse']:.2f}"
    }

    # 2. Key features using permutation importance
    clf_importance = permutation_importance(models["classifier"], X_class, y_class, n_repeats=5, random_state=0)
    reg_importance = permutation_importance(models["regressor"], X_reg, y_reg, n_repeats=5, random_state=0)

    top_clf = pd.Series(clf_importance.importances_mean, index=features).sort_values(ascending=False).head(5)
    top_reg = pd.Series(reg_importance.importances_mean, index=features).sort_values(ascending=False).head(5)

    report["top_classification_features"] = top_clf
    report["top_regression_features"] = top_reg

    # 3. Recommendations
    report["recommendations"] = [
        "Retrain models regularly with recent data.",
        "Use top features for simplified models or explanations.",
        "Track model drift using ROC AUC and RMSE over time."
    ]

    # 4. Visualizations
    plt.figure(figsize=(6, 4))
    top_clf.plot(kind='barh', title='Top 5 Classification Features')
    plt.gca().invert_yaxis()
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(6, 4))
    top_reg.plot(kind='barh', title='Top 5 Regression Features', color='orange')
    plt.gca().invert_yaxis()
    plt.tight_layout()
    plt.show()

    return report

report = create_analysis_report(
    models=optimized_models,
    metrics=improvement_metrics,
    features=X_class.columns,
    X_class=X_class,
    y_class=y_class,
    X_reg=X_reg,
    y_reg=y_reg
)

print("=== Model Performance Summary ===")
print(report["performance_summary"])

print("\n=== Top Classification Features ===")
print(report["top_classification_features"])

print("\n=== Top Regression Features ===")
print(report["top_regression_features"])

print("\n=== Recommendations ===")
for item in report["recommendations"]:
    print(f"- {item}")
```
