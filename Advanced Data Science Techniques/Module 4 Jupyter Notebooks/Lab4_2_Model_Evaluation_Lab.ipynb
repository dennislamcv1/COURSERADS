## Task 1: Initial Setup and Baseline Model Solution Code

<b>Step 1:</b> Environment Setup

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score, roc_curve, auc,
                             confusion_matrix, f1_score, roc_auc_score, precision_recall_curve)
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('insightlysoft_dataset.csv')

# 1. Remove unnecessary variables and encode categorical variables using one-hot encoding
X = df.drop(['is_heavy_user', 'account_id'], axis=1)
X = pd.get_dummies(X, columns=['city', 'gender'], drop_first=True)

# 2. Create train/test split (70% train, 30% test)
# Define the target variable
y = df['is_heavy_user']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

<b>Step 2:</b> Create baseline mode

```python
# Import and initialize the RandomForestClassifier
model = RandomForestClassifier(random_state=42)

# Train the model using the training data
model.fit(X_train, y_train)

# Generate predictions on the test set
y_pred = model.predict(X_test)
y_pred_proba = model.predict_proba(X_test)[:, 1]
```

## Task 2: Implementing Evaluation Metrics Solution Code

<b>Step 1:</b> Create Evaluation Function

```python
# Create a function that calculates and displays multiple metrics
def evaluate_model(y_true, y_pred, y_pred_proba=None):
    """
    Calculate and display model performance metrics

    Parameters:
    y_true: True labels
    y_pred: Predicted labels
    y_pred_proba: Predicted probabilities (optional)

    Returns:
    Dictionary of metrics
    """
    # Your code here to calculate:
    # - Accuracy
    # - Precision
    # - Recall
    # - F1-Score
    # - ROC-AUC (if probabilities provided)
    metrics = {}

    metrics['accuracy'] = accuracy_score(y_true, y_pred)
    metrics['precision'] = precision_score(y_true, y_pred)
    metrics['recall'] = recall_score(y_true, y_pred)
    metrics['f1_score'] = f1_score(y_true, y_pred)

    if y_pred_proba is not None:
        metrics['roc_auc'] = roc_auc_score(y_true, y_pred_proba)

    # Print metrics
    for key, value in metrics.items():
        print(f"{key.capitalize()}: {value:.3f}")

    return metrics

# Test your function
baseline_metrics = evaluate_model(y_test, y_pred)
baseline_metrics = evaluate_model(y_test, y_pred, y_pred_proba)
```

<b>Step 2:</b> Create Visualizations

```python
# Create visualizations to help understand model performance
def plot_model_performance(y_true, y_pred, y_pred_proba=None):
    """
    Create performance visualizations

    Parameters:
    y_true: True labels
    y_pred: Predicted labels
    y_pred_proba: Predicted probabilities (optional)
    """
    plt.figure(figsize=(18, 5))

    # 1. Confusion matrix heatmap
    plt.subplot(1, 3, 1)
    cm = confusion_matrix(y_true, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
    plt.title('Confusion Matrix')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')

    if y_pred_proba is not None:
        # 2. ROC curve
        plt.subplot(1, 3, 2)
        fpr, tpr, _ = roc_curve(y_true, y_pred_proba)
        roc_auc = auc(fpr, tpr)
        plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.2f})')
        plt.plot([0, 1], [0, 1], color='navy', lw=1, linestyle='--')
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title('ROC Curve')
        plt.legend(loc='lower right')
        plt.grid(True)

        # 3. Precision-Recall curve
        plt.subplot(1, 3, 3)
        precision, recall, _ = precision_recall_curve(y_true, y_pred_proba)
        plt.plot(recall, precision, color='green', lw=2)
        plt.xlabel('Recall')
        plt.ylabel('Precision')
        plt.title('Precision-Recall Curve')
        plt.grid(True)
    else:
        # If no probabilities, leave plots 2 and 3 blank or add a message
        plt.subplot(1, 3, 2)
        plt.text(0.5, 0.5, 'No predicted probabilities provided',
                 horizontalalignment='center', verticalalignment='center')
        plt.axis('off')

        plt.subplot(1, 3, 3)
        plt.text(0.5, 0.5, 'No predicted probabilities provided',
                 horizontalalignment='center', verticalalignment='center')
        plt.axis('off')

    plt.tight_layout()
    plt.show()

plot_model_performance(y_test, y_pred, y_pred_proba)
```

## Task 3: Hyperparameter Tuning Solution Code


<b>Step 1:</b> Define Parameter Space

```python
# Define focused parameter space for RandomizedSearchCV
param_distributions = {
    # 1. Number of trees in the forest
    'n_estimators': [50, 100, 150],

    # 2. Maximum depth of the tree
    'max_depth': [None, 10, 20, 30],

    # 3. Minimum samples required to split a node
    'min_samples_split': [2, 5, 10],

    # 4. Minimum samples required at each leaf node
    'min_samples_leaf': [1, 2, 4],
}

# Implement random search
from sklearn.model_selection import RandomizedSearchCV

# 1. Initialize RandomizedSearchCV with focused param grid
random_search = RandomizedSearchCV(
    estimator=RandomForestClassifier(random_state=42),
    param_distributions=param_distributions,  # your defined param grid
    n_iter=10,  # number of parameter settings to try
    scoring='roc_auc',  # metric to optimize
    cv=3,  # 3-fold cross-validation
    random_state=42,
    n_jobs=-1,  # use all processors
    verbose=1
)

# 2. Fit the search on training data
random_search.fit(X_train, y_train)

# 3. Extract best parameters
print("Best parameters found:")
print(random_search.best_params_)
```

<b>Step 2:</b> Evaluate Tuned Model

```python
# Compare baseline and tuned models
# 1. Get predictions from tuned model
best_model = random_search.best_estimator_
y_pred_tuned = best_model.predict(X_test)
y_pred_proba_tuned = best_model.predict_proba(X_test)[:, 1]

# 2. Calculate metrics for both models
print("Baseline model performance:")
baseline_metrics = evaluate_model(y_test, y_pred, y_pred_proba)

print("\nTuned model performance:")
tuned_metrics = evaluate_model(y_test, y_pred_tuned, y_pred_proba_tuned)

# 3. Create comparison visualization: bar chart for key metrics
import matplotlib.pyplot as plt
import numpy as np

metrics_to_compare = ['accuracy', 'precision', 'recall', 'f1_score', 'roc_auc']
baseline_values = [baseline_metrics[m] for m in metrics_to_compare]
tuned_values = [tuned_metrics[m] for m in metrics_to_compare]

x = np.arange(len(metrics_to_compare))
width = 0.35

fig, ax = plt.subplots(figsize=(8, 5))
rects1 = ax.bar(x - width/2, baseline_values, width, label='Baseline')
rects2 = ax.bar(x + width/2, tuned_values, width, label='Tuned')

ax.set_ylabel('Score')
ax.set_title('Model Performance Comparison')
ax.set_xticks(x)
ax.set_xticklabels(metrics_to_compare)
ax.legend()
ax.set_ylim(0, 1.05)  # Slightly increased to prevent label clipping

# Improve label formatting
for rect in rects1 + rects2:
    height = rect.get_height()
    ax.annotate(f'{height:.2f}',  # Keep only 2 decimals
                xy=(rect.get_x() + rect.get_width() / 2, height),
                xytext=(0, 4),  # increase offset for readability
                textcoords="offset points",
                ha='center', va='bottom',
                fontsize=9)  # smaller, cleaner font

plt.tight_layout()
plt.show()
```

## Task 4: Model Selection and Recommendations Solution Code

```python
# Create summary report
def create_model_summary(baseline_metrics, tuned_metrics):
    """
    Create summary comparing model versions

    Parameters:
    baseline_metrics: Dictionary of baseline model metrics
    tuned_metrics: Dictionary of tuned model metrics

    Returns:
    DataFrame comparing models
    """
    summary_df = pd.DataFrame({
        'Baseline': baseline_metrics,
        'Tuned': tuned_metrics
    })
    summary_df['Improvement'] = summary_df['Tuned'] - summary_df['Baseline']
    return summary_df

    return summary_df

summary = create_model_summary(baseline_metrics, tuned_metrics)
print(summary)
```
