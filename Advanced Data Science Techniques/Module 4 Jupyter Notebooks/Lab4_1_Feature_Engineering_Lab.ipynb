## Task 1: Environment Setup and Data Exploration Solution Code

<b>Step 1:</b> Initial Setup

```python
# Import required libraries
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('insightlysoft_dataset.csv')
```

<b>Step 2:</b> Examine the data structure and basic statistics.

```python
# Examine the data structure and basic statistics
# 1. Dataset shape
print("Dataset shape:", df.shape)

# 2. Column data types
print("\nColumn data types:")
print(df.dtypes)

# 3. Basic statistics for numerical columns
print("\nBasic statistics for numerical columns:")
print(df.describe())

# 4. Missing value check
print("\nMissing values per column:")
print(df.isnull().sum())
```

## Task 2: Manual Feature Engineering Solution Code

<b>Step 1:</b> Create Engagement Features

```python
# Create at least 3 new features that capture customer engagement
# Consider combining existing metrics in meaningful ways
# Example structure:
def create_engagement_features(df):
    """
    Create engagement-related features
    Returns DataFrame with new features
    """
     # Total usage: how much time they spend using the product
    df['total_usage'] = df['monthly_login_freq'] * df['avg_session_length']

    # Has been active recently (1 if last activity was within 7 days)
    df['recently_active'] = (df['last_activity_days_ago'] <= 7).astype(int)

    # Interaction level: combines logins and support tickets
    df['interaction_level'] = df['monthly_login_freq'] + df['num_support_tickets']

    return df

# Apply your function
df = create_engagement_features(df)
```

<b>Step 2:</b> Create Customer Health Features

```python
# Create features that might indicate customer health
# Consider:
# - Support ticket patterns
# - Usage trends
# - Satisfaction indicators
def create_health_features(df):
    """
    Create customer health-related features
    Returns DataFrame with new features
    """
    # Number of support tickets per month (normalized by contract length)
    df['tickets_per_month'] = df['num_support_tickets'] / (df['contract_length_months'] + 1e-5)

    # Low usage flag: 1 if avg product usage hours is less than 1 hour (possible risk)
    df['low_usage_flag'] = (df['avg_product_usage_hours'] < 1).astype(int)

    # Satisfaction risk: 1 if customer satisfaction score is below 5 (scale assumed 0-10)
    df['low_satisfaction'] = (df['customer_satisfaction_score'] < 5).astype(int)

    return df

df = create_health_features(df)
```

## Task 3: AI-Assisted Feature Generation Solution Code

<b>Step 1:</b> Implement AI Feature Generator


```python
# Create an AI feature generator that suggests complex feature combinations
def ai_feature_generator(df):
    """
    Simulate AI-suggested features
    Returns DataFrame with new features
    """
    new_features = {}

    # Weighted engagement score combining login frequency, session length, and feature adoption
    new_features['weighted_engagement'] = (
        0.5 * df['monthly_login_freq'] +
        0.3 * df['avg_session_length'] +
        0.2 * df['feature_adoption_score']
    )

    # Interaction between recent activity and customer satisfaction
    new_features['activity_satisfaction_interaction'] = (
        (1 / (df['last_activity_days_ago'] + 1)) * df['customer_satisfaction_score']
    )

    # Ratio of support tickets adjusted by contract length and NPS score
    new_features['tickets_nps_ratio'] = (
        df['num_support_tickets'] / (df['contract_length_months']) * df['nps_score']
    )

    return pd.DataFrame(new_features)


# Generate AI-suggested features
ai_features = ai_feature_generator(df)
df = pd.concat([df, ai_features], axis=1)
```

<b>Step 2:</b> Evaluate AI-Generated Features

```python
# Evaluate the effectiveness of AI-generated features
# 1. Calculate feature importance scores
features = df.drop(['account_id', 'churned', 'city', 'gender'], axis=1, errors='ignore')  # drop non-numeric or target
target = df['churned']

rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(features, target)

importances = rf.feature_importances_
feature_names = features.columns

# Create DataFrame for importance
feat_imp_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importances
}).sort_values(by='Importance', ascending=False)

# 2. Visualization comparing feature importance
plt.figure(figsize=(12, 8))
sns.barplot(data=feat_imp_df.head(15), x='Importance', y='Feature')
plt.title('Top 15 Feature Importances (Including AI-Generated Features)')
plt.xlabel('Importance Score')
plt.ylabel('Feature')
plt.tight_layout()
plt.show()
```

## Task 4: Feature Selection and Optimization Solution Code

<b>Step 1:</b> Analyze Feature Correlations

```python
# Create correlation matrix and visualization
# 1. Calculate correlations between numeric features
numeric_features = df.select_dtypes(include=[np.number])

# Ensure column names are unique (if duplicates exist)
numeric_features = numeric_features.loc[:, ~numeric_features.columns.duplicated()]

correlation_matrix = numeric_features.corr()

# 2. Create correlation heatmap
plt.figure(figsize=(14, 12))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Feature Correlation Heatmap")
plt.show()

# 3. Identify highly correlated feature pairs (absolute correlation > 0.8, excluding self-correlations)
threshold = 0.8

# Select upper triangle of correlation matrix without diagonal
upper_tri = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
)

# Find pairs with correlation greater than threshold
high_corr = upper_tri.stack().loc[lambda x: abs(x) > threshold].sort_values(ascending=False)

print("Highly correlated feature pairs (|corr| > 0.8):")
print(high_corr)
```

<b>Step 2:</b> Select Final Feature Set

```python
# Select optimal feature set based on:
# 1. Importance scores
# 2. Correlation analysis
# 3. Business relevance
def select_optimal_features(df, feat_imp_df, importance_threshold=0.05):
    """
    Select final feature set based on importance and correlation.
    Returns list of selected features
    """
     # Filter features by importance threshold
    important_features = feat_imp_df[feat_imp_df['Importance'] >= importance_threshold]['Feature'].tolist()

    # Compute correlation matrix for important features
    corr_matrix = df[important_features].corr().abs()

    to_drop = set()
    for i in range(len(corr_matrix.columns)):
        for j in range(i):
            if corr_matrix.iloc[i, j] > 0.8:
                f1 = corr_matrix.columns[i]
                f2 = corr_matrix.columns[j]
                # Drop feature with lower importance
                imp_f1 = feat_imp_df.loc[feat_imp_df['Feature'] == f1, 'Importance'].values[0]
                imp_f2 = feat_imp_df.loc[feat_imp_df['Feature'] == f2, 'Importance'].values[0]
                to_drop.add(f1 if imp_f1 < imp_f2 else f2)

    selected_features = [f for f in important_features if f not in to_drop]

    return selected_features

final_features = select_optimal_features(df, feat_imp_df)
print(final_features)
```
