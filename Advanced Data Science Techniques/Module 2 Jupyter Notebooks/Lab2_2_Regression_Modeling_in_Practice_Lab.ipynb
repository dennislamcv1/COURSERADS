## Task 1: Setup and Data Preparation Solution Code

<b>Step 1:</b> Import Libraries and Load Data

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

# Load the preprocessed dataset
df = pd.read_csv('insightlysoft_dataset.csv')
```

## Task 2: Model Building Solution Code

<b>Step 1:</b> Split the Data

```python
# Select only numerical features for your model
features = df.select_dtypes(include=['number']).columns.drop('num_support_tickets')

df_numerical_features = df[features]

# Split data into training and testing sets
# Use 20% for testing
X_train, X_test, y_train, y_test = train_test_split(
    df_numerical_features,
    df['num_support_tickets'],
    test_size=0.2,
    random_state=42
)
```

<b>Step 2:</b> Feature Scaling

```python
# Initialize the StandardScaler to normalize feature distributions
scaler = StandardScaler()

# Fit the scaler on the training data and transform it
X_train_scaled = scaler.fit_transform(X_train)

# Use the same scaler to transform the test data
X_test_scaled = scaler.transform(X_test)
```

<b>Step 3:</b> Create and Train Model

```python
# Initialize and train the model
model = LinearRegression()
model.fit(X_train_scaled, y_train)
```

<b>Step 4:</b> Validation

```python
# Check the shape of your training and testing sets
print(f"X_train shape: {X_train_scaled.shape}")
print(f"X_test shape: {X_test_scaled.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")

# Print model coefficients to see feature importance
# Pair each feature name with its coefficient
for feature, coef in zip(features, model.coef_):
    print(f"{feature}: {coef:.4f}")
```

## Task 3: Model Evaluation Solution Code

<b>Step 1:</b> Make Predictions

```python
# Generate predictions and calculate metrics
y_pred = model.predict(X_test_scaled)

# Calculate RMSE and R²
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"RMSE: {rmse:.2f}")
print(f"R² Score: {r2:.2f}")
```

<b>Step 2:</b> Visualize Results

```python
# Create actual vs predicted plot
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred, alpha=0.5)
plt.plot([y_test.min(), y_test.max()],
         [y_test.min(), y_test.max()],
         'r--', lw=2)
plt.xlabel('Actual Number of Support Tickets')
plt.ylabel('Predicted Number of Support Tickets')
plt.title('Actual vs Predicted Number of Support Tickets')
plt.tight_layout()
plt.show()
```

<b>Step 3:</b> Analyze and Interpret

The RMSE of **0.31** indicates that, on average, the model's predictions differ from the actual **number of support tickets** by just 0.31, which is very low and suggests high accuracy.

The R² score of **1.0** shows that the model explains 100% of the variance in the **number of support tickets**, meaning it captures nearly all the underlying patterns in the data.

In the actual vs. predicted plot, most points lie very close to the diagonal line, confirming that the model performs well across the entire range of values. Only minimal deviations are observed, suggesting few outliers and no major prediction issues.

## Task 4: Business Interpretation Solution Code

<b>Step 1:</b> Analyze Feature Importance

```python
# Create a DataFrame of features and their coefficients
feature_importance = pd.DataFrame({
    'Feature': features,
    'Coefficient': model.coef_
})
print("Feature Importance:")
print(feature_importance.sort_values('Coefficient', ascending=False))
```
